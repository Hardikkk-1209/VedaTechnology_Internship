This notebook demonstrates reusable Python functions for common data-science calculations and transformations.

```python
def calculate_mean(values):
    if not values:
        return 0
    return sum(values) / len(values)

def calculate_maximum(values):
    return max(values) if values else None

def calculate_minimum(values):
    return min(values) if values else None

def calculate_percentage(value, total):
    if total == 0:
        return 0
    return (value / total) * 100

def normalize_values(values):
    if not values:
        return []
    low, high = min(values), max(values)
    if low == high:
        return [0.0] * len(values)
    return [(value - low) / (high - low) for value in values]

def count_above_threshold(values, threshold):
    return sum(1 for value in values if value > threshold)
```

## Function parameters and arguments
A parameter is the name used in a function definition. An argument is the actual value passed when the function is called.

```python
marks = [72, 81, 65, 90, 78, 88, 69]

mean_mark = calculate_mean(marks)
maximum_mark = calculate_maximum(marks)
minimum_mark = calculate_minimum(marks)
pass_percentage = calculate_percentage(count_above_threshold(marks, 40), len(marks))

print("Marks:", marks)
print("Mean:", round(mean_mark, 2))
print("Maximum:", maximum_mark)
print("Minimum:", minimum_mark)
print("Above 40 (%):", round(pass_percentage, 2))
```

```python
sales = [12000, 15000, 11000, 18000, 16000, 14500]

sales_summary = {
    "mean": calculate_mean(sales),
    "maximum": calculate_maximum(sales),
    "minimum": calculate_minimum(sales),
    "normalized": normalize_values(sales)
}

sales_summary
```

## Simple data transformation

```python
def transform_marks(marks, bonus=5):
    return [min(mark + bonus, 100) for mark in marks]

original = [35, 48, 67, 89, 97]
updated = transform_marks(original, bonus=5)

print("Original:", original)
print("After bonus:", updated)
```

## Key takeaways
Functions keep repeated logic in one place, make calculations easier to test, and let the same logic work with different datasets.

## Interview answers
**What is a function in Python?** A reusable block of code that performs a specific task and can accept inputs and return an output.

**Parameter vs argument?** A parameter is a variable in the function definition; an argument is the value supplied to that parameter during a function call.

**Why are functions useful in data science?** They reduce repetition, improve readability, make data processing consistent, simplify testing, and allow the same calculation to be reused across multiple datasets.
