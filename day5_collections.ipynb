This notebook covers Python's four core collection types and practical operations used when handling structured data.

## 1. Lists
Lists are ordered and mutable, so elements can be added, removed and updated.

```python
students = ["Aarav", "Isha", "Rohan", "Meera"]
students.append("Kabir")
students.remove("Rohan")
students[1] = "Ishita"
students.sort()

print("Students:", students)
print("First student:", students[0])
print("Number of students:", len(students))
```

```python
marks = [72, 85, 64, 91, 78]
marks.append(88)
marks.remove(64)
marks.sort()

print("Updated marks:", marks)
print("Highest mark:", marks[-1])
print("Average:", sum(marks) / len(marks))
```

## 2. Tuples
Tuples are ordered but immutable. They are useful for values that should stay fixed.

```python
student_record = (101, "Hardik", "Data Science", 8.6)
print("Student ID:", student_record[0])
print("Name:", student_record[1])
print("Course:", student_record[2])
print("CGPA:", student_record[3])

coordinates = (19.0760, 72.8777)
print("Coordinates:", coordinates)
```

## 3. Sets
Sets store unique values and are useful for removing duplicates and comparing groups of values.

```python
skills = {"Python", "SQL", "Excel", "Python", "Power BI"}
print("Unique skills:", skills)

data_science = {"Python", "SQL", "Statistics", "Pandas"}
analytics = {"SQL", "Excel", "Power BI", "Statistics"}

print("Common skills:", data_science.intersection(analytics))
print("All skills:", data_science.union(analytics))
print("Only data science skills:", data_science.difference(analytics))
```

```python
cities = ["Mumbai", "Pune", "Mumbai", "Nashik", "Pune", "Nagpur"]
unique_cities = set(cities)
print("Original:", cities)
print("Unique cities:", unique_cities)
print("Number of unique cities:", len(unique_cities))
```

## 4. Dictionaries
Dictionaries store key-value pairs and provide direct access to values through their keys.

```python
student = {
    "id": 101,
    "name": "Hardik",
    "course": "Data Science",
    "marks": 86
}

print("Name:", student["name"])
print("Marks:", student["marks"])

student["marks"] = 90
student["grade"] = "A"
student.pop("course")

print("Updated student:", student)
```

```python
employees = {
    "E101": {"name": "Aisha", "department": "Analytics", "salary": 52000},
    "E102": {"name": "Rahul", "department": "Finance", "salary": 48000},
    "E103": {"name": "Neha", "department": "Analytics", "salary": 57000}
}

analytics_employees = [employee["name"] for employee in employees.values() if employee["department"] == "Analytics"]
average_salary = sum(employee["salary"] for employee in employees.values()) / len(employees)

print("Analytics employees:", analytics_employees)
print("Average salary:", average_salary)
```

## Common methods at a glance

```python
example_list = [4, 2, 7]
example_list.append(9)
example_list.remove(2)
example_list.sort()

example_set = {1, 2, 3}
example_set.add(4)
example_set.discard(2)

example_dict = {"name": "Aarav", "score": 82}
example_dict.update({"score": 88, "grade": "A"})

print("List:", example_list)
print("Set:", example_set)
print("Dictionary:", example_dict)
```

## Comparison table

| Collection | Ordered | Mutable | Duplicates | Access | Typical use |
|---|---|---|---|---|---|
| List | Yes | Yes | Allowed | Index | Changing sequences of values |
| Tuple | Yes | No | Allowed | Index | Fixed records |
| Set | No | Yes | Not allowed | Membership | Unique values and comparisons |
| Dictionary | Insertion order preserved | Yes | Keys must be unique | Key | Key-value data |

## Practice with a student dataset

```python
student_data = [
    {"name": "Aarav", "marks": 78, "city": "Mumbai"},
    {"name": "Isha", "marks": 91, "city": "Pune"},
    {"name": "Rohan", "marks": 84, "city": "Mumbai"},
    {"name": "Meera", "marks": 67, "city": "Nashik"}
]

marks = [student["marks"] for student in student_data]
cities = {student["city"] for student in student_data}
top_student = max(student_data, key=lambda student: student["marks"])

print("Average marks:", sum(marks) / len(marks))
print("Cities represented:", cities)
print("Top student:", top_student["name"], top_student["marks"])
```

## Interview answers
**What is the difference between a list, tuple, set, and dictionary?** A list is ordered and mutable, a tuple is ordered and immutable, a set stores unique values, and a dictionary stores key-value pairs.

**Why are sets useful for finding unique values?** Sets automatically remove duplicate values, so converting a collection to a set is a simple way to keep only unique elements.

**How do you access a dictionary value?** Use its key, such as `student["name"]`, or use `.get("name")` when a safer lookup is preferred.

## Key takeaways
Choosing the right collection makes data processing simpler. Lists work well for changing sequences, tuples for fixed records, sets for uniqueness, and dictionaries for structured key-value information.
